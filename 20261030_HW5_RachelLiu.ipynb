```python
import numpy as np
import math

# ---------------------------------------------------------------------------
# PARAMETERS
# with 20 time periods, 4 units, arrival prob of 0.75, 
#   willingness to pay with mean 2
# ---------------------------------------------------------------------------

def initializeData():
    noPeriods_ = 20
    noUnits_ = 4
    arrivalProbs_ = 0.75 * np.ones( noPeriods_ )
    meanWillingness_ = 2 * np.ones( noPeriods_ )
    return( noPeriods_ , noUnits_ , arrivalProbs_ , meanWillingness_ )

# ---------------------------------------------------------------------------
# DEMAND PROBABILITY
# gives e^(-p / meanWillingness) for a given time period
# ---------------------------------------------------------------------------

def demandProb( time , price ) :
    global meanWillingness
    return math.exp( - price / meanWillingness[ time ])

# ---------------------------------------------------------------------------
# OPTIMAL PRICE
# gives value of optimal price at each time period t (stepping backwards) 
#   for each unit amount k
#   by lambda_t * e^(-p_t(k)/mu_t) * mu_t + V_(t+1)(k)
# ---------------------------------------------------------------------------

def computeOptimalPrices():
    global noPeriods , noUnits , arrivalProbs , meanWillingness , values , prices
    for t in range( noPeriods - 1 , -1 , -1 ):
        for k in range( 1, noUnits + 1 , 1):
            price = meanWillingness[ t ] + values[ t + 1 ][ k ] - values[ t + 1 ][ k - 1 ]
            prices[ t ][ k ] = price
            demand = demandProb( t , price )
            value = arrivalProbs[ t ] * demand * meanWillingness[ t ] + values[ t + 1 ][ k ]
            values[ t ][ k ] = value

# ---------------------------------------------------------------------------
# INTIIALIZATION
# creates empty array of values, prices
# ---------------------------------------------------------------------------

( noPeriods , noUnits , arrivalProbs , meanWillingness ) = initializeData()
values = np.zeros( ( noPeriods + 1 , noUnits + 1 ) )
prices = np.zeros( ( noPeriods , noUnits + 1 ) )

# ---------------------------------------------------------------------------
# RUNNING COMPUTATION
# self explanatory
# ---------------------------------------------------------------------------

computeOptimalPrices()
print( prices )
print()
print( values )
```

## Question 2
Consider selling 4 units of a product over 20 time periods. At each time period, there is a
customer arrival with probability 0.5. The willingness to pay of each customer is exponentially
distributed with mean 4.
### (a) Write a program to compute the optimal pricing policy. Submit your commented code.
Your program should work with arbitrary number of units to sell, number of time periods,
customer arrival probabilities and willingness to pay means. You can hardcode the exponential
willingness to pay distribution into your program.

```python
import numpy as np
import math

# ---------------------------------------------------------------------------
# PARAMETERS
# initializeData() takes the number of time periods, number of units,
#   arrival probability and mean willingness to pay (initialized later)
#   and copies the arrival prob and mean into one entry per time period
# ---------------------------------------------------------------------------

def initializeData( noPeriods_ , noUnits_ , arrivalProb_ , meanWillingness_ ):
    arrivalProbs_ = arrivalProb_ * np.ones( noPeriods_ )
    meanWillingness_ = meanWillingness_ * np.ones( noPeriods_ )
    return( noPeriods_ , noUnits_ , arrivalProbs_ , meanWillingness_ )

# ---------------------------------------------------------------------------
# DEMAND PROBABILITY
# probability an arriving customer buys at a price is P(wtp >= p)
#   which for the exponential distribution is e^(-p / meanWillingness)
# ---------------------------------------------------------------------------

def demandProb( time , price ) :
    global meanWillingness
    return math.exp( - price / meanWillingness[ time ])

# ---------------------------------------------------------------------------
# OPTIMAL PRICE
# gives value of optimal price at each time period t (stepping backwards)
#   for each unit amount k
#   optimal price p_t(k) = mu_t + V_(t+1)(k) - V_(t+1)(k-1)
#   value V_t(k) = lambda_t * e^(-p_t(k)/mu_t) * mu_t + V_(t+1)(k)
#   V_T(k) = 0 (unsold units are worth nothing) and V_t(0) = 0 (nothing to sell)
# ---------------------------------------------------------------------------

def computeOptimalPrices():
    global noPeriods , noUnits , arrivalProbs , meanWillingness , values , prices
    for t in range( noPeriods - 1 , -1 , -1 ):
        for k in range( 1, noUnits + 1 , 1):
            price = meanWillingness[ t ] + values[ t + 1 ][ k ] - values[ t + 1 ][ k - 1 ]
            prices[ t ][ k ] = price
            demand = demandProb( t , price )
            value = arrivalProbs[ t ] * demand * meanWillingness[ t ] + values[ t + 1 ][ k ]
            values[ t ][ k ] = value

# ---------------------------------------------------------------------------
# INITIALIZATION
# 20 time periods, 4 units, arrival prob of 0.5, willingness to pay with mean 4
# creates empty array of values, prices
# ---------------------------------------------------------------------------

( noPeriods , noUnits , arrivalProbs , meanWillingness ) = initializeData( 20 , 4 , 0.5 , 4 )
values = np.zeros( ( noPeriods + 1 , noUnits + 1 ) )
prices = np.zeros( ( noPeriods , noUnits + 1 ) )

# ---------------------------------------------------------------------------
# RUNNING COMPUTATION
# prices[ t ][ k ] = optimal price at period t with k units left (column 0 unused)
# values[ t ][ k ] = expected optimal revenue from period t onward with k units left
# ---------------------------------------------------------------------------

computeOptimalPrices()
print( prices )
print()
print( values )
```

### (a) Find interval with width at most 1 containing optimal expected revenue
Consider the case where α1 = 2.1, α2 = 2.4, α3 = 1.8, β1 = 0.1, β2 = 0.2 and β3 = 0.3. Assume that we know that the optimal expected revenue is somewhere in the
interval [0, 60]. Use bisection search to find an interval [a, b] with width at most 1 so that the
optimal expected revenue is guaranteed to be in this interval. (That is, the interval [a, b] needs
to satisfy b − a ≤ 1.) Do all your work by hand for the first 3 iterations of bisection search
and show all your work. You can use a spreadsheet or a program to do the other iterations.

```python
# ---------------------------------------------------------------------------
# BISECTION CHECK FUNCTION
# returns whether sum of e^(alpha_i - 1 - beta_i * t)/beta_i >= t
# ---------------------------------------------------------------------------

def check( alpha1 , alpha2 , alpha3 , beta1 , beta2 , beta3 , t ):
    leftSide = math.exp( alpha1 - 1 - beta1 * t )/beta1 + math.exp( alpha2 - 1 - beta2 * t )/beta2 + math.exp( alpha3 - 1 - beta3 * t )/beta3
    if leftSide >= t:
        return True
    else:
        return False

# ---------------------------------------------------------------------------
# RUNNING COMPUTATION
# returns rounds and interval when interval is less than or equal to 1
# ---------------------------------------------------------------------------

interval = [ 0 , 60 ]
round = 0
while interval[ 1 ] - interval[ 0 ] > 1:
    round += 1
    if check( 2.1 , 2.4 , 1.8 , 0.1 , 0.2 , 0.3 , (interval[ 1 ] + interval[ 0 ])/2 ) == True:
        interval[ 0 ] = (interval[ 1 ] + interval[ 0 ])/2
    else:
        interval[ 1 ] = (interval[ 1 ] + interval[ 0 ])/2
    print( f"after round {round} , this is the new interval: {interval}" )
# print( round , interval )
```
